```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
import seaborn as sns
from tensorflow.keras import layers
```

```python
df= sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.drop(columns=['deck','who'],inplace=True)
```

```python
df['age']=df['age'].fillna(df['age'].mean())
```

```python
df.dropna(subset=['embark_town','embarked'],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.info()
```

```python
obj = df.select_dtypes(include=['object','category']).columns
```

```python
obj
```

```python
df = pd.get_dummies(df,columns=obj,drop_first=True)
```

```python
df.head()
```

```python
bool_col = df.select_dtypes(include='bool').columns
df[bool_col] = df[bool_col].astype(int)
```

```python
df.head()
```

```python
df.drop(columns=['alive_yes'],inplace=True)
```

```python
x = df.drop(columns='survived')
y= df['survived']
```

```python
X_max = x.max()
X_min = x.min()
X_scaled = (x-X_min)/(X_max-X_min + 1e-8)
```

```python
X_scaled.head()
```

```python
X_train,X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.20, random_state=42)
```

```python
model = keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(8,activation='relu'),
    layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='sgd',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=model.fit(X_train.values,y_train.values,validation_data=[X_test.values,y_test.values],epochs=100,batch_size=4, verbose=1)
```
